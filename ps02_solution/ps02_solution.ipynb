# PS02 Solution Notebook
This notebook runs the offline FST + noisy-channel simulator. It does not call an external LLM API.
Run from this folder or update `PROJECT_DIR` to the directory containing `04_fst_noisy_simulator.py`.

```python
from pathlib import Path
import importlib.util
PROJECT_DIR = Path.cwd()
script = PROJECT_DIR / '04_fst_noisy_simulator.py'
if not script.exists():
    script = PROJECT_DIR / 'ps02_solution' / '04_fst_noisy_simulator.py'
spec = importlib.util.spec_from_file_location('ps02_simulator', script)
sim = importlib.util.module_from_spec(spec)
spec.loader.exec_module(sim)
print('Loaded:', script)
```

## Run the ten-turn simulation

```python
transcript = sim.run_demo()
print(transcript)
Path(script).with_name('05_Noisy_Interrogation_Transcript.md').write_text(transcript, encoding='utf-8')
```

## Entropy and channel-capacity reference

```python
import math
print(f'Uniform 7-token entropy: {math.log2(7):.3f} bits')
W, snr = 1.0, 3.0
capacity = W * math.log2(1 + snr)
print(f'Illustrative capacity C = W log2(1+S/N): {capacity:.3f} bits per normalized second')
```
